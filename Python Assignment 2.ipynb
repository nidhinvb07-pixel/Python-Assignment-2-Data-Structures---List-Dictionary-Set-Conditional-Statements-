```python
##List
##1.list creation
age_list = [24, 25, 26, 27, 28]
print(age_list)

name_list = ["Nidhin", "Arun", "Abhirami", "Anu", "Kiran"]
print(name_list)
```

```python
##2.list operations and modification
name_list.append("Yazhini")
print(name_list)
```

```python
age_list.insert(2, 30)
print(age_list)
```

```python
name_list.remove("Yazhini")
print(name_list)
```

```python
age_list.pop()
print(age_list)
```

```python
age_list.extend([29, 30, 26])
print(age_list)
```

```python
age_list.sort(reverse=True)
print(age_list)
```

```python
print("Maximum age:", max(age_list))
print("Minimum age:", min(age_list))
print("Sum of all ages:", sum(age_list))
```

```python
##3.accesing list elements
print(name_list[0])
```

```python
print(name_list[-1])
```

```python
print(name_list[2:5])
```

```python
print(name_list[::-1])
```

```python
##sets(operations)
my_set = {'a', 'e', 'i', 'o', 'u', 'a', 'a', 'i'}
print(my_set)
##A set does not allow duplicate values.
##The values 'a' and 'i' are repeated, but they appear only once in the output.
##Sets are unordered, so the output order may vary.
```

```python
my_set[4] = 's'
##sets are mutable, but since it is unordered, it do not support indexing.Therefore, we cannot access or modify elements using an index.
```

```python
set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}

print(set1)
print(set2)
```

```python
##union
print(set1.union(set2))
```

```python
##intersection
print(set1.intersection(set2))
```

```python
##operators & conditional statements(if,elif,else)
score = float(input("Enter your score (0 to 10): "))

if score < 0 or score > 10:
    print("Invalid score! Please enter a score between 0 and 10.")

elif score > 7:
    print("Above Average: Excellent performance! Keep it up!")

elif score >= 4:
    print("Average: Good effort! Keep practicing, there's room for improvement.")

else:
    print("Below Average: Need to improve your performance. Consistent practice will lead to better results.")
```
